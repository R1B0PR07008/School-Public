```python
import pandas as pd
```

```python
excel = pd.read_excel('Imaginary Business Clients.xlsx')
```

```python
excel
```

```python
basic : float = 0
Standard : float = 0
Premium : float = 0
vip : float = 0

basic_ : float = 0
Standard_ : float = 0
Premium_ : float = 0
vip_ : float = 0

for i in excel["Membership_Type"]:

    column = excel["Membership_Type"].tolist().index(i)

    if i == "Basic":
        basic += float(excel["Total_Spent_USD"][column])
        basic_ += 1
    elif i == "Standard":
        Standard += float(excel["Total_Spent_USD"][column])
        Standard_ += 1
    elif i == "Premium":
        Premium += float(excel["Total_Spent_USD"][column])
        Premium_ += 1
    elif i == "VIP":
        vip += float(excel["Total_Spent_USD"][column])
        vip_ += 1

basic = basic / basic_
Standard = Standard / Standard_
Premium = Premium / Premium_
vip = vip / vip_

print("Basic: " + str(basic) + "\n" + str(basic_))
print("Standard: " + str(Standard))
print("Premium: " + str(Premium))
print("VIP: " + str(vip))

if basic > Standard and basic > Premium and basic > vip:
    print("Basic is the most popular membership type.")
elif Standard > basic and Standard > Premium and Standard > vip:
    print("Standard is the most popular membership type.")
elif Premium > basic and Premium > Standard and Premium > vip:
    print("Premium is the most popular membership type.")
elif vip > basic and vip > Standard and vip > Premium:
    print("VIP is the most popular membership type.")
```
